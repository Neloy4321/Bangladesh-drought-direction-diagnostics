# Paper 2 — reviewer evidence audit
This notebook checks the publicly shared **completed synthetic benchmark summaries**. It does not re-fit the original 15,600 cases or use external field records. For raw-level verification, supply the separate Kaggle output ZIP and run `scripts/verify_results.py --kaggle-zip <path>`.


```python
from pathlib import Path
import csv, subprocess, sys
repo = Path.cwd().resolve()
if not (repo/'scripts'/'verify_results.py').exists():
    repo = repo.parent
assert (repo/'scripts'/'verify_results.py').exists(), 'Start Jupyter in the repository or its notebooks/ folder'
subprocess.run([sys.executable, str(repo/'scripts'/'verify_results.py')],cwd=repo,check=True)
```

```python
with (repo/'results'/'official'/'Official_Final_Summary.csv').open(newline='',encoding='utf-8') as f:
    rows=list(csv.DictReader(f))
print('Method / scenario summary rows:',len(rows))
for method in ['ccm','pcmci','granger','conditional_lag_ols']:
    sample=[r for r in rows if r['method']==method]
    print(method, 'attempted',sum(int(r['attempted']) for r in sample),
          'eligible',sum(int(r['eligible']) for r in sample),
          'detected',sum(int(r['detected']) for r in sample))
```
